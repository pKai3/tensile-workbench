## Views and settings
- Property tables above the plot controls show per-specimen data, group statistics, Instron comparisons and source/fit checks. **Export tables only** does not generate plots.
- Graph management, sample checkboxes, legend labels and colour overrides apply to the whole graph. Set colours under **Labels and colours** and click **Apply group overrides**.
- Each plot selector has a **Settings** dropdown beneath it. Titles are view-specific; axis limits and calculation parameters shared by related views remain linked and are labelled accordingly.
- **Plot display** switches between the unchanged static SVG viewer and interactive Plotly. It does not recalculate curves or change averages. The choice is saved per graph.
- Static plots retain click-to-enlarge and image zoom. Plotly supports hover values, drag-to-zoom, panning, legend toggling and an **Enlarge plot** button. Its numbered individual-curve labels are display indices, not specimen IDs.
- Zooming or hiding curves in Plotly is temporary exploration, not specimen exclusion. Publication exports retain the configured axes and all calculated curves.
- **Export plot set** writes the existing 600-DPI static plots and optional Excel tables only when explicitly requested. Renderer choice does not change these outputs.
- Definitions autosave in `tensile_workbench_v18.project.json`. First launch copies your latest v17 definitions; subsequent edits in the two versions are independent.
- Restart the kernel after software updates. The environment remains outside OneDrive; the launcher stays beside the notebook.
